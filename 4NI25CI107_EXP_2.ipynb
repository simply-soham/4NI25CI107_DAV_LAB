Exploratory Data Analysis: Loading datasets, computing descriptive statistics, identifying missing values and duplicates.


```python
# ------------------------------------------------------------
# 1. IMPORT PANDAS
# ------------------------------------------------------------
import pandas as pd
# ------------------------------------------------------------
# 2. LOAD THE DATASET
# ------------------------------------------------------------
# If the CSV file is in the same folder as your Jupyter Notebook:
df = pd.read_csv("student_performance.csv")

print("Dataset loaded successfully!")
# ------------------------------------------------------------
# 3. DISPLAY FIRST 5 ROWS
# -----------------------------------------------------------
print("\nFirst 5 rows:")
print(df.head())
# ------------------------------------------------------------
# 4. DISPLAY LAST 5 ROWS
# ------------------------------------------------------------
print("\nLast 5 rows:")
print(df.tail())
# ------------------------------------------------------------
# 5. NUMBER OF ROWS AND COLUMNS
# -----------------------------------------------------------
print("\nShape of the dataset:")
print(df.shape)
# ------------------------------------------------------------
# 6. DISPLAY COLUMN NAMES
# ----------------------------------------------------------
print("\nColumn names:")
print(df.columns)
# ------------------------------------------------------------
# 7. DISPLAY DATA TYPES
# ------------------------------------------------------------
print("\nData types:")
print(df.dtypes)


# ============================================================
# EDA - EXPLORATORY DATA ANALYSIS
# ============================================================
# ------------------------------------------------------------
# 8. DESCRIPTIVE STATISTICS
# ------------------------------------------------------------
print("\nDescriptive Statistics:")
print(df.describe())

# ------------------------------------------------------------
# 9. DESCRIPTIVE STATISTICS INCLUDING CATEGORICAL DATA
# ------------------------------------------------------------
print("\nStatistics including categorical columns:")
print(df.describe(include='all'))

# ------------------------------------------------------------
# 10. CHECK MISSING VALUES
# ------------------------------------------------------------
print("\nMissing values in each column:")
print(df.isnull().sum())


# ------------------------------------------------------------
# 11. CHECK DUPLICATE ROWS
# ------------------------------------------------------------
print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

# ------------------------------------------------------------
# 12. CHECK DUPLICATE STUDENT IDs
# ------------------------------------------------------------
print("\nNumber of duplicate student IDs:")
print(df['student_id'].duplicated().sum())	

# ------------------------------------------------------------
# 13. DISPLAY GRADE COUNTS
# ------------------------------------------------------------
print("\nNumber of students in each grade:")
print(df['grade'].value_counts())

# ============================================================
# DATA WRANGLING
# ============================================================

# ------------------------------------------------------------
# 15. CREATE A NEW DATAFRAME FOR MERGING
# ------------------------------------------------------------
grade_info = pd.DataFrame({
    'grade': ['A', 'B', 'C', 'D', 'F'],
    'grade_description': [
        'Excellent',
        'Good',
        'Average',
        'Below Average',
        'Fail'
    ]
})

print("\nGrade Information DataFrame:")
print(grade_info)

# ------------------------------------------------------------
# 16. MERGING TWO DATAFRAMES
# ------------------------------------------------------------
merged = pd.merge(
    df,
    grade_info,
    on='grade',
    how='left'
)

print("\nMerged DataFrame:")
print(merged.head())

# ============================================================
# JOIN
# ============================================================

# ------------------------------------------------------------
# 17. CREATE DATAFRAME FOR JOIN
# ------------------------------------------------------------
grade_details = pd.DataFrame({
    'description': [
        'Excellent',
        'Good',
        'Average',
        'Below Average',
        'Fail'
    ],
    'performance_level': [
        'Very High',
        'High',
        'Medium',
        'Low',
        'Very Low'
    ]
}, index=['A', 'B', 'C', 'D', 'F'])

# ------------------------------------------------------------
# 18. COUNT STUDENTS BY GRADE
# ------------------------------------------------------------
grade_counts = df['grade'].value_counts().to_frame('student_count')

# ------------------------------------------------------------
# 19. JOIN TWO DATAFRAMES
# ------------------------------------------------------------
joined = grade_counts.join(grade_details)

print("\nJoined DataFrame:")
print(joined)

# ============================================================
# CONCATENATION
# ============================================================

# ------------------------------------------------------------
# 20. SPLIT DATAFRAME INTO TWO PARTS
# ------------------------------------------------------------
df1 = df.iloc[:100]
df2 = df.iloc[100:200]

# ------------------------------------------------------------
# 21. CONCATENATE THE TWO DATAFRAMES
# ------------------------------------------------------------
combined = pd.concat([df1, df2])

print("\nConcatenated DataFrame:")
print(combined.head())

print("\nShape after concatenation:")
print(combined.shape)
```
